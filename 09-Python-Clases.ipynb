<h2 id="creating">Crear una Clase</h2>

Creemos ahora una clase circle, pero primero, importemos una librería para dibujar objetos:

```python
# Importar la librería

import matplotlib.pyplot as plt
%matplotlib inline
```

El primer paso en la creación de nuestra clase es usar la palabra reservada <code>class</code>, y después el nombre de la clase como se muestra en la figura 4. En este curso la clase padre será siempre objeto:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%203/Images/ClassesDefine.png" width="400" />

<i>Figura 4: Tres instancias de la clase circle o, dicho de otra forma, tres objetos de tipo circle.</i>

El siguiente paso es el método especial llamado constructor <code>&#95;&#95;init&#95;&#95;</code>, que es usado para inicializar el objeto. La entrada son datos para los atributos. El termino <code>self</code> contiene todos los atributos del grupo. Por ejemplo <code>self.color</code> asigna el valor del atributo color y <code>self.radius</code> valor del radio. También tenemos el método <code>add_radius()</code> con el parámetro <code>r</code>, el método suma el valor de <code>r</code> al atributo radius. Para acceder al valor de radius usamos la sintaxis <code>self.radius</code>. El etiquetado se resume en la Figura 5.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%203/Images/ClassesCircle.png" width="600" />

<i>Figura 5: Sintaxis del objeto circle.</i>

El objeto actual se muestra mas adelante. Incluimos el método <code>drawCircle</code> para mostrar la imagen de un circulo. Asignamos el valor de radius en 3 y el color en azul (blue):

```python
# Crear la clase Circle

class Circle(object):

    # Constructor
    def __init__(self, radius=3, color='blue'):
        self.radius = radius
        self.color = color

    # Method
    def add_radius(self, r):
        self.radius = self.radius + r
        return(self.radius)

    # Method
    def drawCircle(self):
        plt.gca().add_patch(plt.Circle((0, 0), radius=self.radius, fc=self.color))
        plt.axis('scaled')
        plt.show()
```

<div style="background-color:#eef7ff;padding:10px;border-left:4px solid #2b7de9;">

**Nota personal (Luis):** Una clase se declara con `class Nombre(object):` y adentro se define el constructor `__init__`, que es el método que se ejecuta automáticamente al crear un objeto nuevo y sirve para inicializar sus atributos.

</div>

<h2 id="rect">La Clase Rectángulo</h2>

Vamos a crear una clase rectángulo con los atributos height (altura) y color. Solo añadiremos el método para dibujar un rectángulo:

```python
# Crear una nueva clase Rectangle para construir un objeto rectangulo

class Rectangulo(object):

    # Constructor
    def __init__(self, width=2, height=3, color='r'):
        self.height = height
        self.width = width
        self.color = color

    # Method
    def drawRectangle(self):
        plt.gca().add_patch(plt.Rectangle((0, 0), self.width, self.height ,fc=self.color))
        plt.axis('scaled')
        plt.show()
```

##**Actividad a realizar:**


La clase matplotlib.patches.Ellipse, permite dibujar elipses desde [matplotlib](https://matplotlib.org/stable/api/_as_gen/matplotlib.patches.Ellipse.html). La clase se importa de la siguiente manera:

*from matplotlib.patchs import Ellipse*

Se puede dibujar entonces cualquier elipse, dando los argumentos:

plt.gca().add_patch(Ellipse(xy, width, height, facecolor = color, edgecolor = edgecolor))

Se requiere crear una clase Elipse, la cual el constructor sea con valores iniciales cualesquiera. Adicionalmente, la clase debe tener los siguientes métodos:

1.- Cambiar el ancho

2.- Cambiar el alto

3.- Cambiar el color de relleno

4.- Cambiar el color de borde

5.- Dibujar

<div style="background-color:#eef7ff;padding:10px;border-left:4px solid #2b7de9;">

**Nota personal (Luis):** Repetir el mismo patrón (constructor + método para dibujar) con una clase distinta (Rectangulo en vez de Circle) me ayudó a confirmar que la estructura de una clase siempre es la misma: constructor, atributos, y los métodos que se necesiten según lo que el objeto tenga que hacer.

</div>

<div style="background-color:#eef7ff;padding:10px;border-left:4px solid #2b7de9;">

**Desarrollo de la actividad (Luis):** Sigo el mismo patrón usado para `Circle` y `Rectangulo`: un constructor con valores por defecto, un atributo por cada característica de la elipse (posición, ancho, alto, color de relleno y color de borde) y un método por cada acción que pide el enunciado (cambiar ancho, cambiar alto, cambiar color de relleno, cambiar color de borde y dibujar).

</div>

```python
# Actividad: Clase Elipse
# Importo Ellipse desde matplotlib.patches (el enunciado tenía un typo: 'patchs' en vez de 'patches')
from matplotlib.patches import Ellipse
import matplotlib.pyplot as plt
%matplotlib inline


class Elipse(object):
    # Constructor con valores iniciales por defecto
    def __init__(self, xy=(0, 0), width=4, height=2, facecolor='blue', edgecolor='black'):
        self.xy = xy
        self.width = width
        self.height = height
        self.facecolor = facecolor
        self.edgecolor = edgecolor

    # 1.- Cambiar el ancho
    def set_width(self, width):
        self.width = width

    # 2.- Cambiar el alto
    def set_height(self, height):
        self.height = height

    # 3.- Cambiar el color de relleno
    def set_facecolor(self, facecolor):
        self.facecolor = facecolor

    # 4.- Cambiar el color de borde
    def set_edgecolor(self, edgecolor):
        self.edgecolor = edgecolor

    # 5.- Dibujar
    def drawEllipse(self):
        plt.gca().add_patch(Ellipse(self.xy, self.width, self.height,
                                     facecolor=self.facecolor, edgecolor=self.edgecolor))
        plt.axis('scaled')
        plt.show()
```

```python
# Pruebo la clase Elipse: creo una instancia, cambio sus atributos con los métodos y la dibujo
MyEllipse = Elipse(xy=(0, 0), width=6, height=3, facecolor='green', edgecolor='black')
MyEllipse.set_width(8)
MyEllipse.set_height(4)
MyEllipse.set_facecolor('orange')
MyEllipse.set_edgecolor('purple')
MyEllipse.drawEllipse()
```
